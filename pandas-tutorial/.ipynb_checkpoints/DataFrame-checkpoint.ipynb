# Creating a DataFrame

```python
data={
      'Name':['john','anna','pater','linda'],
      'Age' :[45,56,23,43],
      'City':['jatni','bbsr','burla','banglore'],
      'salary':[50000,60000,35000,100000]
}
df=pd.DataFrame(data)
```

```python
df
```

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
data2=data_list
df2=pd.DataFrame(data2)
df2
```

```python
col=['Name','Age','City','Salary']
df2=pd.DataFrame(data2,columns=col)
df2
```
